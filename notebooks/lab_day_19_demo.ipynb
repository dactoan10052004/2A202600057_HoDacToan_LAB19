```python
flat_engine = FlatRAGEngine(client=client, answer_model=CONFIG.answer_model, embedding_model=CONFIG.embedding_model)
flat_engine.build_index(chunks)
flat_engine.answer('Who founded OpenAI?', top_k=5)
```
